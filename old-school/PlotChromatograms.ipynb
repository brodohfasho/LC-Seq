```python
## Create chromatograms from sequencing count data for 10-40%B data
## Control image quality at the bottom

# Import packages as easy abbreviations
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from IPython.display import display
import os
%matplotlib inline
from tqdm.auto import tqdm

# Define excel sheet and workbook
excelsheet = "/Users/grant/Library/CloudStorage/GoogleDrive-brodohfasho@gmail.com/Shared drives/DEL Lipophilicity Paper/Data Spreadsheets/10-40_Testset.xlsx"
workbook = "Sheet1"

# Define an output folder for .png's
output_folder = "/Users/grant/Desktop/Test/"
os.makedirs(output_folder, exist_ok=True)

# Create pandas dataframe from sequencing data
# Define indexes ("Common_Name (N-->C), lid")
df = pd.read_excel(excelsheet, sheet_name=workbook)
df.set_index(["Common_Name (N-->C)", "lid"], inplace=True)

# Group by the first level of the MultiIndex
grouped_df = df.groupby(level=0)
out_dict = {}

# Iterate over each group in the grouped dataframe
for index, group_df in tqdm(grouped_df):
    plt.figure(figsize=(10, 6))  # Increase figure size for better clarity
    handles = []  # Collect handles for legend
    labels = []   # Collect labels for legend
    for (lid, row) in group_df.iterrows():
        
        # Define new dataframe of delimited RT_count data
        chrom_df = pd.Series(row["all_datapoints"])
        
        # Split the comma-delimited values into a new dataframe (time, counts) 
        chromvalues_df = chrom_df.str.split(",", expand=True)
        chromvalues_new = chromvalues_df.transpose()
        master_df = chromvalues_new[0].str.split("[:,;]", expand=True)
        
        # Filter out "None" values
        # Convert split delimited data from strings to numerical "float" values for plotting
        # Divide time points by 60 s to convert to minutes
        Time = [float(value) / 60 for value in master_df[0] if value is not None] 
        Scaled_count = [float(value) for value in master_df[2] if value is not None]
        
        # Sort Time and Scaled_count based on Time values
        sorted_data = sorted(zip(Time, Scaled_count))
        Time, Scaled_count = zip(*sorted_data)
        
        # Customize the appearance of each line
        color = 'darkgrey' if 'DEL-0044' in lid else 'navy'  # Use more distinct colors
        label = 'Linear' if 'DEL-0044' in lid else 'Cyclized'
        
        # Plot the data as a scatter plot with a connecting line
        scatter = plt.scatter(Time, Scaled_count, label=label, color=color, alpha=0.8, marker='o', s=10)  # Increase marker size
        line, = plt.plot(Time, Scaled_count, linestyle='-', linewidth=2, color=color, alpha=1)  # Increase line width
        
        # Add handles and labels to the lists
        handles.append(line)
        labels.append(label)
        
    # Manually reorder the legend labels to ensure 'Linear' is on top
    sorted_indices = sorted(range(len(labels)), key=lambda k: (labels[k] != 'Linear', k))
    handles = [handles[i] for i in sorted_indices]
    labels = [labels[i] for i in sorted_indices]
   
    # Customize legend and other plot properties
    plt.legend(handles, labels, loc='upper left', fontsize=12, frameon=True, fancybox=True)  # Increase font size and add frame and box to legend
    plt.xlim(0, 65)
    plt.ylim(0, plt.gca().get_ylim()[1])
    plt.tick_params(axis='both', which='major', labelsize=12)  # Increase tick label size
    plt.ticklabel_format(style='plain', axis='y')
    plt.xlabel('Time (min)', fontsize=14)  # Increase font size for labels
    plt.ylabel('Scaled_counts', fontsize=14)
    plt.title(f"{index}", fontsize=16)  # Increase font size for title
    plt.tight_layout()  
    plot_filename = os.path.join(output_folder, f'{index}.png')
    plt.savefig(plot_filename, dpi=300, bbox_inches='tight')  # Increase DPI and save with tight bounding box
    plt.close()
```
